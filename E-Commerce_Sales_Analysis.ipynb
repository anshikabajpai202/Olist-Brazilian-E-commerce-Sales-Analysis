## Data Cleaning Decisions

1. No duplicate rows were found in the datasets checked.

2. Order timestamp columns were converted from string format to datetime formaT.

3. Missing order approval and delivery dates were retained..

4. The 610 products with missing product category names were assigned the category "Unknown".

5. Missing product physical attributes were retained.

6. Missing review titles and review messages were retained because written comments are not required for the review-score distribution analysis. The review_score column contains no missing values.

## 6. SALES ANALYSIS

Q1.Which product category has the highest revenue?

```python
category_revenue = sales_data.groupby("product_category_name")["revenue"].sum()
category_revenue = category_revenue.sort_values(ascending=False)
```

```python
category_revenue
```

```python
top_category = category_revenue.index[0]
print("Top Category:", top_category)
top_revenue = category_revenue.iloc[0]
print("Highest Revenue:", top_revenue)
```

### Finding

The **beleza_saude** category generated the highest revenue, with total product revenue of approximately **1.26 million**. 

### Q2. Which month had peak sales?

```python
order["order_purchase_timestamp"]
```

```python
sales_data["month"] = sales_data["order_purchase_timestamp"].dt.month
```

```python
sales_data[["order_purchase_timestamp", "month"]].head()
```

```python
monthly_sales = sales_data.groupby("month")["revenue"].sum()
monthly_sales = monthly_sales.sort_index()
monthly_sales
```

```python
peak_month = monthly_sales.idxmax()
peak_sales = monthly_sales.max()

print("Peak Month:", peak_month)
print("Peak Sales:", peak_sales)
```

```python
import calendar
```

```python
peak_month_name = calendar.month_name[peak_month]

print("Peak Month:", peak_month_name)
print("Peak Sales:", peak_sales)
```

### Finding

May recorded the highest sales, with total product revenue of **1,502,588.82**. This was the peak sales month in the dataset based on the calculated monthly revenue.
