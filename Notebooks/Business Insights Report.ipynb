# 1. Business Insights

### 1.1: Overall Sales Performance

- The dataset contains approximately **98.7K unique orders**.
- Total revenue generated is approximately **13.59M**.
- The Average Order Value (AOV) is approximately **137.75**.
- This indicates a large volume of transactions with a significant overall sales contribution.

### 1.2. Top Product Category

- **Beleza_Saude** is the highest-revenue product category.
- It generated approximately **1.26M** in revenue.
- This category should be considered an important contributor to overall business revenue.

### 1.3. Monthly Sales Performance

- **May** was the highest-performing month.
- Revenue in May was approximately **1.50M**.
- January had approximately **1.07M** in revenue, while September had the lowest monthly revenue at approximately **0.62M**.
- Sales performance varies considerably across months.

### 1.4. Regional Sales Performance

- **São Paulo (SP)** was the best-performing state.
- It generated approximately **5.20M** in revenue.
- This shows that SP is a major regional market for the business.

### 1.5. Customer Review Analysis

- **5-star reviews** were the most common review score.
- There were **57,328** 5-star reviews.
- 4-star reviews were the second most common with **19,142** reviews.
- This indicates that a large proportion of customers gave positive ratings.

### 1.6. Order Status

- The majority of orders were marked as **delivered**.
- There were **96,478 delivered orders**.
- Shipped orders were the next largest group with **1,107 orders**.
- Canceled and unavailable orders were relatively small compared with delivered orders.

# 2. Business Recommendations

### 2.1. Focus on High-Revenue Categories

The business should continue investing in high-performing categories such as **Beleza_Saude** and other top-revenue categories.

### 2.2. Analyze Seasonal Sales Patterns

Since **May** recorded the highest revenue, the business can study the factors behind this peak and use similar strategies during lower-performing months.

### 2.3. Strengthen Regional Strategy

Since **SP** generated the highest revenue, the company can maintain strong operations in this market while identifying opportunities to increase sales in lower-performing states.

### 2.4. Maintain Customer Satisfaction

The high number of 5-star reviews indicates positive customer satisfaction. The business should maintain product and service quality to preserve this performance.

### 2.5. Investigate Low-Performing Months

Months such as **September** showed lower revenue. The company can investigate possible reasons and consider targeted promotions or marketing campaigns during weaker periods.

# 3: Conclusion

This project analyzed e-commerce sales data to understand overall sales performance, customer behavior, product categories, monthly trends, regional performance, order values, and customer reviews.

The analysis showed that **Beleza_Saude** was the highest-revenue product category, while **May** was the highest-performing month. **São Paulo (SP)** was the top-performing state by revenue, and the majority of customer reviews were **5-star ratings**.

The analysis and Power BI dashboard provide useful insights into sales trends, high-performing categories, regional markets, and customer satisfaction. These insights can help businesses identify strong markets, understand sales patterns, and make data-driven decisions.
