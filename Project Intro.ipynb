# E-Commerce Sales Analysis

## 1. Project Introduction

This project analyzes e-commerce sales data to understand sales performance, product category performance, monthly sales trends, regional performance, order values, and customer reviews.

The objective is to clean and transform the raw data, perform exploratory data analysis using Python, identify important business insights, and create an interactive dashboard using Power BI.

The analysis focuses on understanding sales patterns and identifying areas that can support data-driven business decisions.

## 2. Objectives

The main objectives of this project are:

- Analyze overall sales performance.
- Identify the highest-revenue product categories.
- Analyze monthly sales trends.
- Identify the best-performing states/regions.
- Analyze order value distribution.
- Analyze customer review scores.
- Identify important business insights.
- Create an interactive Power BI dashboard.
- Provide business recommendations based on the analysis.

## 3. Tools & Technologies

- **Python** – Data cleaning, transformation, analysis, and visualization
- **Pandas** – Data manipulation and analysis
- **NumPy** – Numerical operations
- **Matplotlib** – Data visualization
- **Seaborn** – Statistical visualization
- **Power BI** – Interactive dashboard and reporting
- **Jupyter Notebook** – Project development and documentation

## 4. Dataset Description

The project uses multiple e-commerce datasets containing information about:

- Orders
- Order items
- Products
- Customers
- Customer reviews

These datasets are combined during the data preparation process to create a sales dataset for analysis.

The project follows these steps:
